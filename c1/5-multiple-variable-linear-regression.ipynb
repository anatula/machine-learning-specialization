### Model Prediction with Multiple Variables
The model's prediction with multiple variables is given by the linear model:

$$ f_{\mathbf{w},b}(\mathbf{x}) =  w_0x_0 + w_1x_1 +... + w_{n-1}x_{n-1} + b \tag{1}$$ $$ f_{\mathbf{w},b}(\mathbf{x}) = \mathbf{w} \cdot \mathbf{x} + b  \tag{2} $$ 


```python
# Equation (1) computes the prediction for one single training example (one row of X_train)
#         x0     x1   x2   x3        y
# X[0] = [2104   5    1    45]       y[0] = 460
# X[1] = [1416   3    2    40]       y[1] = 232
# X[2] = [ 852   2    1    35]       y[2] = 178

# X[0] = [2104, 5, 1, 45] → the features of example 0
# y[0] = 460 → the true answer for example 0
```

```python
def predict_single_loop(x,w,b):
    total = 0
    for i in range(len(x)):
        total += w[i] * x[i]
        print(f"i={i}, w[{i}]+x[{i}]={total}") 
    return total + b
```

```python
# using row 0 of X_train
f_wb = predict_single_loop(X_train[0], w_init, b_init)
print(f"f_wb shape {f_wb.shape}, prediction: {f_wb}")
```

### Cost with Multiple Variables


The equation for the cost function with multiple variables $J(\mathbf{w},b)$ is:
$$J(\mathbf{w},b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})^2 \tag{3}$$ 
where:
$$ f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = \mathbf{w} \cdot \mathbf{x}^{(i)} + b  \tag{4} $$ 






```python
def compute_cost(X, y, w, b):
    m = X.shape[0]
    cost = 0.0
    for i in range(m):
        f_wb_i = np.dot(X[i],w)+b
        cost += (f_wb_i-y[i])**2
        print(cost)
    cost = cost / (2*m)
    return cost
```

### Gradient descent with multiple variables

$$\begin{align*} \text{repeat}&\text{ until convergence:} \; \lbrace \newline\;
& w_j = w_j -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial w_j} \tag{5}  \; & \text{for j = 0..n-1}\newline
&b\ \ = b -  \alpha \frac{\partial J(\mathbf{w},b)}{\partial b}  \newline \rbrace
\end{align*}$$

where, n is the number of features, parameters $w_j$,  $b$, are updated simultaneously and where  

$$
\begin{align}
\frac{\partial J(\mathbf{w},b)}{\partial w_j}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})x_{j}^{(i)} \tag{6}  \\
\frac{\partial J(\mathbf{w},b)}{\partial b}  &= \frac{1}{m} \sum\limits_{i = 0}^{m-1} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)}) \tag{7}
\end{align}
$$
* m is the number of training examples in the data set

    
*  $f_{\mathbf{w},b}(\mathbf{x}^{(i)})$ is the model's prediction, while $y^{(i)}$ is the target value

```python
def compute_derivate_w(X,y,w,b):
    m,n = X.shape
    derivate_vector = np.zeros(n)
    for j in range(n):
        res = 0
        for i in range(m):
            f_wb_i = np.dot(X[i],w)+b
            res += (f_wb_i-y[i])*X[i][j]
        derivate = res / m
        derivate_vector[j] = derivate
    return derivate_vector

def compute_derivate_b(X,y,w,b):
    m,n = X.shape
    res = 0
    for i in range(m):
        f_wb_i = np.dot(X[i],w)+b
        res += (f_wb_i-y[i])
    derivate_b = res / m
    return derivate_b

def compute_gradient(X,y,w,b):
    m,n = X.shape
    dj_dw = np.zeros((n,))
    dj_db = 0.
    for i in range(m):
        err = (np.dot(X[i],w)+b) - y[i]
        # ( (X[0][0] * w[0] + X[0][1] * w[1] + X[0][2] * w[2] + X[0][3] * w[3]) + b ) - y[0]
        for j in range(n):
            # j=0: err * X[0][0]  → Size   of house 0
            # j=1: err * X[0][1]  → Bedrooms of house 0
            # j=2: err * X[0][2]  → Floors of house 0
            # j=3: err * X[0][3]  → Age    of house 0
            dj_dw[j] += err * X[i][j]
        dj_db = dj_db + err
    dj_dw = dj_dw / m                                
    dj_db = dj_db / m 
    return dj_db, dj_dw
```

```python
print(compute_derivate_w(X_train, y_train, w_init, b_init))
print(compute_derivate_b(X_train, y_train, w_init, b_init))
print(compute_gradient(X_train, y_train, w_init, b_init))
```

```python
def gradient_descent(X, y, w_in, b_in, cost_fuction, gradient_function, alpha, num_iters):
    w = copy.deepcopy(w_in)  #avoid modifying global w within function
    b = b_in
    for i in range(num_iters):
        dj_db,dj_dw = gradient_function(X, y, w, b)
        w = w - alpha * dj_dw               
        b = b - alpha * dj_db
    return w, b
```

```python
initial_w = np.zeros_like(w_init)
initial_b = 0.
# some gradient descent settings
iterations = 1000
alpha = 5.0e-7
# run gradient descent 
w_final, b_final = gradient_descent(X_train, y_train, initial_w, initial_b, compute_cost, compute_gradient, alpha, iterations)
print("w_final", w_final)
print("b_final", b_final)
```
