NumPy library: extends Python's base capabilities. Add a richer data set: more numeric types, vectors, matrices, and many matrix functions.
NumPy and Python work together fairly seamlessly: Python arithmetic operators work on NumPy data types and many NumPy functions will accept Python data types.

Dimention overload:
- vector: number of elements
- numpy arrays: number of indexes of an array. 1D, has one index
In Course 1 we'll represent vectors as Numpy 1D arrays: `1D array: shape(n,) -> n elements indexed [0] through [n-1]`

```python
a = np.zeros(4);                print(f"np.zeros(4) :   a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.zeros((4,));             print(f"np.zeros(4,) :  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.random_sample(4); print(f"np.random.random_sample(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
```

```python
# https://www.youtube.com/watch?v=pgPwaNpC0so
# reshape changes shape of array without altering its data
# pass number of rows and columns you want .reshape(rows,columns)
# returns a new array
array = np.arange(1,13)
print(array)
print(array.shape)

array = array.reshape(2,6) # need to have matching elements, 2 rows, 6 columns
print(array.shape)
array = array.reshape(3,4) # 3 rows, 4 columns
print(array)
array = array.reshape(4, 3) # 4 rows, 3 columns
print(array)
array = array.reshape(1,12) # 1 rows, 12 columns
print(array)
```

```python
# add a dimention .reshape(layers, rows, columns)
array = array.reshape(2,2,3) # 3D array (like image 3)
print(array)
print('-- -- -- --')
array = array.reshape(3,2,2)
print(array)
```

```python
# set one dimention to be -1: numpy will automatically infer the correct size for that dimention based on the number of columns
t = np.arange(1,13)
t = t.reshape(-1,1) # 1 row, 1 column
print(t)
```

```python
t = t.reshape(1,-1) # 2D 1 row, 12 columns
print(t)
```
