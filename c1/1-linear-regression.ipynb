```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0, 2.0])
y_train = np.array([300.0, 500.0])
```

```python
m = x_train.shape[0]
print(f"Number of training examples is: {m}")
```

$$ f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{1}$$

```python
def compute_output_model(x, w, b):
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = x[i] * w + b
    return f_wb 
```

```python
plt.scatter(x_train, y_train)
plt.title("House Prices")
plt.xlabel("Size in 1000 sqft units")
plt.ylabel("Price in 1000 of $")

w = 100

b = 100
f_wb = compute_output_model(x_train, w, b)
plt.plot(x_train, f_wb)

w = 150
f_wb = compute_output_model(x_train, w, b)
plt.plot(x_train, f_wb)

w = 200
f_wb = compute_output_model(x_train, w, b)
plt.plot(x_train, f_wb)

x_sq_meters = 1.2
predict_price_y_hat = w * x_sq_meters + b
plt.scatter(x_sq_meters, predict_price_y_hat)
```
